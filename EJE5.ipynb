Se coloca una rata en el laberinto que se muestra :    

En ausencia de aprendizaje, cabría suponer que la rata se desplaza por el laberinto de forma aleatoria; es decir, si existen $k$ posibles salidas desde un compartimento, la rata elige cada una de ellas con probabilidad $1/k$. En cada unidad de tiempo, la rata se traslada a un compartimento adyacente. Sea $X_n$ el compartimento ocupado en el instante $n$. El compartimento 7 contiene comida y en el 8 hay un mecanismo de descargas eléctricas. ¿Cuál es la probabilidad de que la rata, al moverse aleatoriamente, encuentre la comida antes de recibir una descarga?

como en el laverinto tenemos la oportunidad de llegar a $1$ y a $2$  desde $0$ entonces tenemos $P(x=1|x=0)=1/2$ y $P(x=0|x=2)=1/2$

```python
import networkx as nx
import matplotlib.pyplot as plt

G = nx.Graph()

# Añadir los nodos
G.add_nodes_from([0, 1, 2])

#(a la arista (0,1) y (0, 2) le damos el valor de 0.5)
G.add_edge(0, 1, weight=1/2) 
G.add_edge(0, 2,weight=1/2)

# Configurar el diseño para poder pintar las etiquetas de las aristas
pos = nx.spring_layout(G)

# Dibujar el grafo básico
nx.draw(G, pos, with_labels=True, node_color="#F4ABAA", node_size=1000)

# Dibujar los valores/pesos sobre las líneas
labels = nx.get_edge_attributes(G, 'weight')
nx.draw_networkx_edge_labels(G, pos, edge_labels=labels)

plt.savefig("GraphUndirected.png", dpi=600)
```
